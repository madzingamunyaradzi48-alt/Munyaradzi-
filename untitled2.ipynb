```python
import matplotlib.pyplot as plt

# 1. Your client data
names = ["Chipo", "Tinashe", "Farai", "Ruvimbo", "Blessing", "Munyaradzi"]
scores = [720, 550, 680, 510, 790, 450]

# 2. Assign colors: Green for safe (>=600), Red for high-risk (<600)
colors = ["green" if s >= 600 else "red" for s in scores]

# 3. Build and display the chart
plt.figure(figsize=(7, 4))
plt.bar(names, scores, color=colors)
plt.axhline(y=600, color="black", linestyle="--", label="Risk Threshold (600)")

plt.title("Bank Client Credit Risk Assessment")
plt.ylabel("Credit Score")
plt.legend()
plt.show()
```
